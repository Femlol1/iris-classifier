```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
x = iris.data #shape(150,4)
y = iris.target #shape(159,)
print(iris.feature_names, iris.target_names)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:5])
print(y_test[:5])
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report
classification_report(y_test, y_pred)
```

```python
from sklearn.tree import plot_tree
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 10))
plot_tree(model, filled=True)
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train, y_train)
y_pred_knn = model_knn.predict(X_test)
print("k-NN accuracy:", accuracy_score(y_test,y_pred_knn))
```

```python
## Results Summary

The model performs very well on the Iris dataset, correctly classifying the test samples with high accuracy. The decision tree and k-NN models both achieve strong results, indicating that the flower measurements are highly informative for distinguishing the three species. The evaluation metrics confirm that the classifier generalizes well to unseen data.
```
